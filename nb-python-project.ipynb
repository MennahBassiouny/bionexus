## 1. Fetch a real gene from NCBI
NCBI asks for your email so they can contact you if there is a problem. Use your real address.

```python
from Bio import Entrez, SeqIO

Entrez.email = "your.email@example.com"   # use your own email

handle = Entrez.efetch(db="nucleotide", id="NM_000207", rettype="fasta", retmode="text")
record = SeqIO.read(handle, "fasta")
handle.close()

seq = str(record.seq)
print(record.description)
print("Length:", len(seq), "bases")
```

## 2. Base composition

```python
tally = {}
for base in seq:
    tally[base] = tally.get(base, 0) + 1

print(tally)
for base, n in tally.items():
    print(base, n, round(n / len(seq) * 100, 1), "%")
```

## 3. GC content

```python
def gc_content(s):
    return round((s.count("G") + s.count("C")) / len(s) * 100, 1)

print("GC content:", gc_content(seq), "%")
```

## 4. From gene to protein
The mRNA starts with an untranslated region, so find the first start codon (ATG) and translate from there.

```python
from Bio.Seq import Seq

start = seq.find("ATG")
print("Start codon at position", start)

cds = seq[start:]
cds = cds[:len(cds) // 3 * 3]          # keep whole codons only
protein = Seq(cds).translate(to_stop=True)
print("Protein length:", len(protein), "amino acids")   # 110: preproinsulin
print("First 20:", protein[:20])
```

## 5. Visualize it

```python
import matplotlib.pyplot as plt

bases = ["A", "C", "G", "T"]
heights = [tally.get(b, 0) for b in bases]

plt.bar(bases, heights, color=["#5b9cf8", "#a855f7", "#10b981", "#f59e0b"])
plt.title("Base composition of human insulin (NM_000207)")
plt.ylabel("Count")
plt.show()
```
