We'll store a DNA sequence and ask it a few simple questions. Your results should match the lesson exactly.

```python
# Lines 1-3: store the sequence, show it, and measure it
dna = "ATGCGTACGTTAGCCGTA"
print(dna)
print(len(dna))
```

```python
# Lines 4-7: count each base
print("A:", dna.count("A"))
print("C:", dna.count("C"))
print("G:", dna.count("G"))
print("T:", dna.count("T"))
```

```python
# Lines 8-10: GC content, the percentage of bases that are G or C
gc = (dna.count("G") + dna.count("C")) / len(dna) * 100
print("GC content:", round(gc, 1), "%")
print("AT content:", round(100 - gc, 1), "%")
```
